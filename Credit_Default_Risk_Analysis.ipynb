```python
df = pd.read_excel(
    "/content/default of credit card clients.xls",
    skiprows=1)
print("Dataset loaded successfully!")
print("Shape:", df.shape)
df.head()
```

```python
df.columns = (df.columns.str.strip()
    .str.replace(" ", "_")
    .str.replace(".", "_", regex=False))

df = df.rename(columns={"PAY_0": "PAY_1",
    "default_payment_next_month": "DEFAULT",
    "default_payment_next_month_": "DEFAULT"})

print(df.columns.tolist())
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
print("Unique customer IDs:", df["ID"].nunique())
```

```python
analysis_df = df.drop(columns=["ID"]).copy()
```

```python
print(analysis_df["DEFAULT"].value_counts(dropna=False))
```

What percentage of customers defaulted on their next payment?

```python
total_customers = len(analysis_df)

defaulted = (analysis_df["DEFAULT"] == 1).sum()
not_defaulted = (analysis_df["DEFAULT"] == 0).sum()

default_rate = defaulted / total_customers * 100

print("Total customers:", total_customers)
print("Customers who defaulted:", defaulted)
print("Customers who did not default:", not_defaulted)
print(f"Default rate: {default_rate:.2f}%")
```

```python
counts = analysis_df["DEFAULT"].value_counts().sort_index()
plt.figure(figsize=(7, 5))
plt.bar(
    ["No default", "Default"],
    [counts.get(0, 0), counts.get(1, 0)])
plt.title("Distribution of Credit Default Outcomes")
plt.xlabel("Payment outcome")
plt.ylabel("Number of customers")
plt.tight_layout()
plt.show()
```


### Interpretation

The dataset contains 30000 customer records, of which 6636 customers defaulted on their next payment and 23364 customers did not.

The overall default rate was 22% in the dataset. This means that approximately 22 out of every 100 customer records were associated with a default on the next payment.

The bar chart shows the distribution of default and non-default outcomes. These results provide an overview of the target variable and help us understand the proportion of default cases before investigating the factors associated with default.

This analysis describes historical data and does not establish the causes of default.


Do customers with delayed repayments have different default rates?

```python
repayment_risk = (
    analysis_df.groupby("PAY_1")["DEFAULT"]
    .agg(["count", "mean"])
    .reset_index())
repayment_risk["default_rate_pct"] = (
    repayment_risk["mean"] * 100)
print(repayment_risk)
```

```python
plt.figure(figsize=(9, 5))
plt.bar(
    repayment_risk["PAY_1"].astype(str),
    repayment_risk["default_rate_pct"])
plt.title("Default Rate by September Repayment Status")
plt.xlabel("Repayment status (PAY_1)")
plt.ylabel("Default rate (%)")
plt.tight_layout()
plt.show()
```

Observations
- Repayment-status categories 3 and 7 show high default rates, approximately 76% and 78%, respectively.
- Categories 2, 3, and 4 also have relatively high default rates.
- Categories -2, -1, and 0 have lower default rates, around 12–17%.
- Default rates do not increase consistently across every category.
- Overall, repayment status appears to be associated with the likelihood of default in the following month.

Does the observed default rate differ across credit-limit ranges?

```python
analysis_df["CREDIT_LIMIT_GROUP"] = pd.cut(
    analysis_df["LIMIT_BAL"],
    bins=[0, 50000, 100000, 200000, 500000, float("inf")],
    labels=["Up to 50k","50k–100k","100k–200k",
            "200k–500k","Above 500k"],
    include_lowest=True)

limit_analysis = (analysis_df.groupby(
        "CREDIT_LIMIT_GROUP",
        observed=False)["DEFAULT"].agg(["count", "mean"])
        .reset_index())
limit_analysis["default_rate_pct"] = limit_analysis["mean"] * 100

print(limit_analysis)
```

```python
plt.figure(figsize=(9, 5))
plt.bar(
    limit_analysis["CREDIT_LIMIT_GROUP"].astype(str),
    limit_analysis["default_rate_pct"])

plt.title("Default Rate by Credit Limit")
plt.xlabel("Credit-limit group (NTD)")
plt.ylabel("Default rate (%)")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

Does default rate vary by age group?

```python
analysis_df["AGE_GROUP"] = pd.cut(
    analysis_df["AGE"],
    bins=[0, 25, 35, 45, 55, 100],
    labels=["18–25", "26–35", "36–45", "46–55", "56+"])

age_analysis = (
    analysis_df.groupby("AGE_GROUP", observed=False)["DEFAULT"]
    .agg(["count", "mean"]))
age_analysis["default_rate_pct"] = age_analysis["mean"] * 100

print(age_analysis)
```

```python
age_analysis["default_rate_pct"].plot(
    kind="bar",
    figsize=(8, 5),
    title="Default Rate by Age Group",
    ylabel="Default rate (%)",
    xlabel="Age group")

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

Examine correlations

```python
columns = [
    "LIMIT_BAL",
    "AGE",
    "PAY_1",
    "BILL_AMT1",
    "PAY_AMT1",
    "DEFAULT"]
correlation = analysis_df[columns].corr()
plt.figure(figsize=(8, 6))
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f")
plt.title("Correlation Between Selected Variables")
plt.tight_layout()
plt.show()
```

# Credit Default and Risk Analysis Using Python

## 1. Objective
The objective of this project was to analyze historical credit-card customer data and explore patterns associated with default payments.

## 2. Dataset
Source: UCI Machine Learning Repository — Default of Credit Card Clients.

The dataset contains customer credit limits, repayment status, bill amounts, payment amounts, age and the next-month default outcome.

## 3. Tools Used
- Python
- Pandas and NumPy
- Matplotlib and Seaborn
- Scikit-learn (optional machine-learning extension)

## 4. Methodology
The project involved data inspection, cleaning, descriptive statistics, exploratory data analysis, visualization and comparisons of default rates across customer groups.

## 5. Key Findings
- Overall default rate: 22%.
- Repayment-status group with the highest observed default rate: Category 7.
- Credit-limit group with the highest observed default rate: Up to 50,000 NTD .
- Age-group comparison: The 56+ age group had the highest default rate at approximately 27%, while the 26–35 age group had the lowest at approximately 20%.

## 6. Limitations
The data is historical and comes from credit-card customers in Taiwan. Observed associations do not prove causation, and the results should not be generalized automatically to current customers or other countries.

## 7. Conclusion
The analysis demonstrates how Python and statistical techniques can be used to explore financial data, communicate patterns and develop an introductory understanding of credit risk.
